```python
# 필요한 라이브러리 불러오기

import pandas as pd                  # 판다스: 표 형태의 데이터 처리
import numpy as np                   # 넘파이: 숫자 및 배열 계산
import matplotlib.pyplot as plt      # 맷플롯립: 그래프 그리기
from sklearn.preprocessing import MinMaxScaler  # 민맥스 스케일러: 데이터 크기 조정
from tensorflow.keras.models import Sequential  # 시퀀셜: 층을 순서대로 쌓는 모델
from tensorflow.keras.layers import LSTM, Dense  # LSTM: 순서가 있는 데이터 학습 / Dense: 결과 계산
```

```python
# 데이터 불러오기

data_path = 'dataset/airline.csv'  # 데이터 경로
df = pd.read_csv(data_path)         # CSV 파일을 읽어서 데이터프레임으로 저장

# data_path(데이터 패스): 데이터 파일이 있는 위치
# read_csv(리드 씨에스브이): CSV 파일을 읽는 기능
# df(디에프): 불러온 표 형태의 데이터
```

```python
# 데이터 확인
print(df.head())
```

```python
# 날짜(Date)와 탑승객 수(count) 열의 이름 설정 및 변환

df['date'] = pd.to_datetime(df['date'])  # 날짜 데이터를 날짜 형식으로 변환
df.set_index('date', inplace=True)       # 날짜를 데이터의 기준(인덱스)으로 설정

# to_datetime(투 데이트타임): 글자로 된 날짜를 날짜 형식으로 변환
# set_index(셋 인덱스): 지정한 열을 데이터의 기준으로 설정
# inplace=True(인플레이스 트루): 원본 데이터에 바로 적용
# 햄버거 가게의 일일 주문 장부에서 날짜를 기준으로 주문 내역을 정렬하는 것
```

```python
# 데이터 시각화

plt.figure(figsize=(12, 6))  # 그래프 크기 설정 (가로 12, 세로 6)
plt.plot(df['count'], label='Passenger Count')  # 월별 탑승객 수를 선 그래프로 표시
plt.title('Monthly Airline Passenger Numbers')  # 그래프 제목
plt.xlabel('Date')  # X축: 날짜
plt.ylabel('Passengers')  # Y축: 탑승객 수
plt.legend()  # 범례 표시
plt.show()  # 그래프 출력

# figure(피겨): 그래프 도화지 만들기
# figsize(피겨 사이즈): 그래프 가로·세로 크기
# plot(플롯): 데이터를 선 그래프로 그리기
# label(레이블): 그래프에 표시할 이름
# xlabel / ylabel: X축과 Y축 이름
# legend(레전드): 범례 표시
# show(쇼): 완성된 그래프 보여주기
# 햄버거 가게의 월별 주문량 변화를 그래프로 확인하는 것
```

```python
# 데이터 정규화

scaler = MinMaxScaler(feature_range=(0, 1))  # 0~1 사이로 변환
scaled_data = scaler.fit_transform(df[['count']])  # 탑승객 수 변환

# MinMaxScaler: 숫자 크기 조정 도구
# feature_range: 변환 범위
# fit_transform: 기준을 찾고 변환
# scaled_data: 변환된 데이터
# 🍔 제각각인 햄버거 가격을 0~1 점수로 바꾸기
```

```python
# 데이터셋 분리 (학습 데이터와 테스트 데이터)

train_size = int(len(scaled_data) * 0.8)  # 전체 데이터의 80%
train_data = scaled_data[:train_size]     # 앞쪽 80% 학습용
test_data = scaled_data[train_size:]      # 뒤쪽 20% 테스트용

# len: 데이터 개수
# int: 정수로 변환
# 🍔 과거 주문으로 공부하고, 이후 주문으로 시험 보기
```

```python
# 시계열 데이터셋 생성 함수

def create_dataset(data, time_step=12):  # 과거 12개월 기준
    X, y = [], []  # 입력값과 정답
    for i in range(len(data) - time_step):  # 데이터 이동
        X.append(data[i:i+time_step])  # 과거 12개월 저장
        y.append(data[i+time_step])  # 다음 달 정답 저장
    return np.array(X), np.array(y)  # 배열로 변환

# create_dataset: 학습 데이터 만들기
# time_step: 참고할 과거 기간
# X: 과거 12개월 / y: 다음 달 정답
# append: 목록에 추가
# 🍔 지난 12개월 주문량으로 다음 달 주문량 맞히기
```

```python
time_step = 12  # 과거 12개월로 다음 달 예측

X_train, y_train = create_dataset(train_data, time_step)  # 학습용 입력값과 정답 생성
X_test, y_test = create_dataset(test_data, time_step)      # 테스트용 입력값과 정답 생성

# X_train: 학습용 과거 12개월 데이터
# y_train: 학습용 다음 달 정답
# X_test: 테스트용 과거 12개월 데이터
# y_test: 테스트용 다음 달 정답
# 🍔 과거 주문 장부로 연습 문제와 시험 문제 만들기
```

```python
# LSTM 입력에 맞게 데이터 형태 변경

X_train = X_train.reshape(X_train.shape[0], X_train.shape[1], 1)  # 학습 데이터 모양 변경
X_test = X_test.reshape(X_test.shape[0], X_test.shape[1], 1)      # 테스트 데이터 모양 변경

# reshape(리셰이프): 데이터 모양 바꾸기
# shape[0]: 데이터 묶음 수
# shape[1]: 과거 12개월 수
# 1: 탑승객 수 데이터 1종류
# 🍔 주문표를 LSTM 요리사가 보기 편하게 3차원으로 정리
```

```python
# LSTM 모델 생성
model = Sequential([
    LSTM(50, return_sequences=True, input_shape=(time_step, 1)),  # 이전 LSTM에 결과 전달
    LSTM(50, return_sequences=False),  # 최종 정보 정리
    Dense(25),  # 중간 계산
    Dense(1)    # 다음 달 탑승객 수 예측
])

# Sequential(시퀀셜): 층을 순서대로 연결
# LSTM(엘에스티엠): 과거 흐름을 기억하는 층
# 50: 기억 단위 50개
# return_sequences: 다음 층에 결과 전달 여부
# input_shape: 과거 12개월 데이터 입력
# Dense(덴스): 정보를 모아 계산하는 층
# 🍔 두 요리사가 지난 주문을 분석하고, 마지막 직원이 다음 달 주문량을 예측
```

```python
# 모델 컴파일
model.compile(optimizer='adam', loss='mean_squared_error')

# compile(컴파일): AI가 공부할 방법 정하기
# optimizer(옵티마이저): 틀린 답을 고치는 방법
# adam(아담): 틀린 부분을 알아서 조정
# loss(로스): 정답과 얼마나 다른지 확인
# mean_squared_error: 틀린 정도를 제곱해 평균 내기
# 🍔 햄버거 맛을 평가하고, 레시피를 고치는 방법 정하기
```

```python
# 모델 학습
history = model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=50,       # 전체 데이터 50회 학습
    batch_size=32,   # 32개씩 묶어서 학습
    verbose=1,       # 학습 진행 상황 표시
    shuffle=False    # 시계열 순서 유지
)
```

```python
# 학습 과정 시각화
plt.figure(figsize=(12, 6))  # 그래프 크기 설정
plt.plot(history.history['loss'], label='Training Loss')  # 학습 오차
plt.plot(history.history['val_loss'], label='Validation Loss')  # 검증 오차
plt.title('Model Loss')  # 그래프 제목
plt.xlabel('Epochs')  # 학습 반복 횟수
plt.ylabel('Loss')  # 오차
plt.legend()  # 범례 표시
plt.show()  # 그래프 출력

# history(히스토리): 학습 기록
# loss(로스): 학습 데이터의 오차
# val_loss(밸 로스): 검증 데이터의 오차
# Epoch(에폭): 전체 데이터를 학습한 횟수
# 햄버거를 연습할수록 맛 평가 점수가 어떻게 변하는지 확인
```

```python
# 테스트 데이터 예측

train_predict = model.predict(X_train)  # 학습 데이터 예측
test_predict = model.predict(X_test)    # 테스트 데이터 예측

# predict(프리딕트): 학습한 모델로 결과 예측
# train_predict: 학습 데이터 예측 결과
# test_predict: 테스트 데이터 예측 결과
# 햄버거 요리사가 연습 문제와 시험 문제의 답을 각각 내놓는 것
```

```python
# 예측값 역정규화

train_predict = scaler.inverse_transform(train_predict)  # 학습 예측값을 원래 숫자로 복원
y_train = scaler.inverse_transform(y_train.reshape(-1, 1))  # 학습 정답값을 원래 숫자로 복원
test_predict = scaler.inverse_transform(test_predict)  # 테스트 예측값을 원래 숫자로 복원
y_test = scaler.inverse_transform(y_test.reshape(-1, 1))  # 테스트 정답값을 원래 숫자로 복원

# inverse_transform(인버스 트랜스폼): 정규화했던 숫자를 원래 값으로 되돌림
# reshape(리셰이프): 데이터 모양 변경
# -1: 행 개수는 자동으로 계산
# 1: 열을 1개로 맞춤
# 햄버거 가격을 0~1로 줄여 계산했다면, 다시 실제 가격으로 되돌리는 과정
```

```python
# 예측 결과 시각화

plt.figure(figsize=(12, 6))  # 그래프 크기 설정

plt.plot(df.index[:len(y_train)], y_train, label='Actual Train Data')  # 실제 학습 데이터
plt.plot(df.index[time_step:len(train_predict) + time_step], train_predict, label='Train Prediction')  # 학습 데이터 예측값

plt.plot(df.index[len(train_data):len(train_data) + len(test_predict)], y_test, label='Actual Test Data')  # 실제 테스트 데이터
plt.plot(df.index[len(train_data):len(train_data) + len(test_predict)], test_predict, label='Test Prediction')  # 테스트 데이터 예측값

plt.title('Actual vs Predicted')  # 실제값과 예측값 비교
plt.xlabel('Date')  # 가로축: 날짜
plt.ylabel('Passengers')  # 세로축: 탑승객 수
plt.legend()  # 범례 표시
plt.show()  # 그래프 출력

# Actual(액추얼): 실제값
# Predicted(프리딕티드): 예측값
# index(인덱스): 날짜 데이터의 위치
# len(렌): 데이터 개수
# 햄버거 가게의 실제 판매량과 AI가 예상한 판매량을 날짜별로 비교하는 그래프야.
```
