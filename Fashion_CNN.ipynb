```python
import tensorflow as tf #로빈 , 그 버거를 하루에 천 개씩 자동으로 찍어내는 최첨단 주방 기계
from tensorflow import keras # 키오스크 기계가져오기 
import numpy as np # 넘파이 숫자계산
import matplotlib.pyplot as plt # 플레이팅
```

```python
# 데이터  불러오기

fashion_mnist = keras.datasets.fashion_mnist #패션 이미지 데이터 창고

# 데이터 학습용,실험용 나누기
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()
# train_images = 학습할 옷 사진 (X). train_labels = 학습할 옷 사진의 정답 (Y)
# test_images = 시험 볼 옷 사진 (X) .test_labels = 시험 볼 옷 사진의 정답 (Y)
```

```python
# 전처리 이미지필셀 값을 0~1사이로 학습용,시험용 둘다 

train_images = train_images / 255.0
test_images = test_images /255.0
```

```python
# 데이터 모양 확인
print("학습용 이미지 모양:", train_images.shape)
print("학습용 라벨 모양:", train_labels.shape)
print("시험용 이미지 모양:", test_images.shape)
print("시험용 라벨 모양:", test_labels.shape)
```

```python
# cnn  차원추가
#  흑백 사진에 채널 추가
train_images = np.expand_dims(train_images, axis = -1) #학습용흑백채널
test_images = np.expand_dims(test_images, axis=-1) #시험용흑백채널
```

```python
# 모델만들기 사진판독 주방설계 뼈대 
model = keras.Sequential ([
    keras.Input(shape=(28,28,1)),
    keras.layers.Conv2D(32, (3,3), activation='relu'), 
    # 컨블루션 (합성곱) 사진의 모서리,모양 특징을 찾는탐정 32명
    # (3,3) 3x3크기로 필터 
    #렐루 음수는 휴지통에 버리고 휴지통 완전삭제 
    # 입력사진크기는 28x28 흑백사진 1채널
    keras.layers.MaxPooling2D((2,2)),
    # 맥스풀링 큰숫자 하나만 남기기 압축 
    keras.layers.Flatten(),
    # 평탄화 list 만들기 
    keras.layers.Dense(128, activation='relu'),
    # 덴스층 , 편처진 사진을 조합해 옷종류 단서 찾는거
    # 128 계산담당자 128개 음수는0, 양수는통과!
    keras.layers.Dense(10, activation='softmax')
])
    # 출력층, 10가지 종류 후보10개의점수를 합계 100% 되도록 정리 
    
    
```

```python
# 모델컴파일 학습규칙 정하기
model.compile(
    optimizer='adam',#눈치빠른 주방장 
    loss='sparse_categorical_crossentropy', # 채점 
    metrics=['accuracy'] # 정확도 얼마나 맞혔는지 측정 
)
```

```python
# 모델 훈련
history = model.fit(
    train_images, train_labels, #학습용ㅇ 사진,정답 공부시키기
    epochs=5, # 5번 반복 공부시키기
    validation_split=0.2 # 검증용으로 학습데이터 20$ 따로 중간 성적확인 
)
```

```python
# 모델 평가  최종 성적 확인
test_loss, test_acc = model.evaluate(
    test_images, test_labels # 학습에 사용ㅎ아지 않은 시험용 사진과 정답 
)

print("테스트 손실:", test_loss) # ai 가 얼마나 틀렸는지
print("테스트 정확도:", test_acc) # ai 가 얼마나 맞혔는지
```
